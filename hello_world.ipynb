```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph  # framework that helps you design and manage the flow of the tasks in your application using a graph structure
```

```python
# We now create an AgentState - Shared data structure that keeps track of information as your application runs.

class AgentState(TypedDict):  # from TypedDict, Python knows AgentState is actually a dictionary under the hood, not a traditional object instance.
    message : str 

def greeting_node(state: AgentState) -> AgentState:
    """Simple node that adds a greeting message to the state """

    state['message'] = "Hey" + state['message'] + ", how is your day going?"

    return state
```
